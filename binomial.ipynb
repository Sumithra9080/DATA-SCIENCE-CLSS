```python
from scipy.stats import binom as b
import numpy as np
import matplotlib.pyplot as plt
```

```python
#The formula for the binomial distribution is
#(P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}\),
#which calculates the probability of getting exactly k successes in n independent trials
```

```python
n=10
p=0.6
x=np.arange(0,n+1)
pmf=b.pmf(x,n,p)
pmf
```

```python
plt.figure(figsize=(8,5))
plt.bar(x,pmf)
plt.show()
```

```python
cdf=b.cdf(x,n,p)
cdf
```

```python
plt.figure(figsize=(8,5))
plt.bar(x,cdf)
plt.show()
```

```python
#The Poisson distribution formula is
#\(P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}\),
#which calculates the probability of a specific number of events (k) occurring in a fixed interval
```

```python
from scipy.stats import poisson as p
```

```python
l=10
k=15
pmf=p.pmf(k,l)
x=np.arange(0,21)
fit=p.pmf(x,l)
fit
```

```python
plt.figure(figsize=(8,5))
plt.bar(x,fit)
plt.xlabel("Number of Success")
plt.ylabel("Probability percent")
plt.title("Poisson Distribution")
plt.show()
```

```python
#CI
from scipy import stats 
salary=[100000,250000,450000,6700000,65400000,5345778,3500000,653277767,34500000]
sample_mean=np.mean(salary)
CI=stats.t.interval(
    confidence=0.95,
    df=len(salary)-1,
    loc=sample_mean,
    scale=stats.sem(salary)

)
print(f"this is my confidence leve{CI}")
```
