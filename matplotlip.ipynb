```python
import matplotlib.pyplot as plt
```

```python
df.groupby("Sleep Hours")[["Hours Studied","Previous Scores"]].mean().plot(kind="bar")
plt.title("Students performance")
plt.xlabel("hours")
plt.ylabel("scores")
plt.legend()
plt.show()
```
