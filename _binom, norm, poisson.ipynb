```python
data = df["Age"].dropna()

mean = data.mean()
std = data.std()

x = np.linspace(data.min(), data.max(), 100)
y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.xlabel("Delivery Time")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

```python
data = df["latitude"].dropna()

lam = data.mean()

x = np.arange(0, int(data.max()) + 1)
y = poisson.pmf(x, lam)

plt.bar(x, y)
plt.xlabel("Number of Orders")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

```python
#binomial distribution 
data = df["longitude"].dropna()

p = (data == "Yes").mean()
n = 10

x = np.arange(0, n + 1)
y = binom.pmf(x, n, p)

plt.bar(x, y)
plt.xlabel("Number of Successful Deliveries")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```
