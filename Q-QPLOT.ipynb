```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as ss
```

```python
data = np.random.normal(loc=50,scale=3,size=100)
data
```

```python
ss.probplot(data,dist="norm",plot=plt)
```

```python
data1 = np.random.exponential(scale=3,size=100)
data1
```

```python
ss.probplot(data1,dist="expon",plot=plt)
```

```python
data1 = np.random.uniform(loc=1,size=100)
data1
```
